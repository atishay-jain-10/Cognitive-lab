```python
scores = (10, 0, 20, 40, 10, 70, 0, 40)
print("Scores Tuple:", scores)
```

```python
highest = max(scores)
highest_index = scores.index(highest)
print("\nHighest Score:", highest)
print("Index of Highest Score:", highest_index)
```

```python
lowest = min(scores)
lowest_count = scores.count(lowest)
print("Lowest Score:", lowest)
print("Number of times Lowest Score appears:", lowest_count)
```

```python
reversed_list = list(scores[::-1])
print("\nReversed Tuple as List:", reversed_list)
```

```python
search = int(input("\nEnter a score to search: "))

if search in scores:
    print("First occurrence index:", scores.index(search))
else:
    print("Score not present in the tuple.")
```

```python
try:
    scores[0] = 100
except TypeError as e:
    print("\nError:", e)
```

```python
first_score, second_score, *remaining_scores = scores
print("\nFirst Score:", first_score)
print("Second Score:", second_score)
print("Remaining Scores:", remaining_scores)
```

```python
import random

random.seed(1024170358)

numbers = [random.randint(100, 900) for _ in range(100)]
print("Random List:")
print(numbers)
```

```python
odd_numbers = [num for num in numbers if num % 2 != 0]
print("\nNumber of Odd Numbers:", len(odd_numbers))
print("Odd Numbers:")
print(odd_numbers)
#Count and print all even numbers
even_numbers = [num for num in numbers if num % 2 == 0]
print("\nNumber of Even Numbers:", len(even_numbers))
print("Even Numbers:")
print(even_numbers)
#Count and print all prime numbers Function to check prime
def is_prime(n):
if n < 2:
return False
for i in range(2, int(n**0.5) + 1):
if n % i == 0:
return False
return True
```

```python
prime_numbers = [num for num in numbers if is_prime(num)]

print("\nNumber of Prime Numbers:", len(prime_numbers))
print("Prime Numbers:")
print(prime_numbers)
```

```python
most_frequent = max(set(numbers), key=numbers.count)
frequency = numbers.count(most_frequent)
print("\nMost Frequent Number:", most_frequent)
print("Frequency:", frequency)
```

```python
digits = [1, 0, 2, 4, 1, 7, 0, 4, 1, 4]
# Create sets A and B
A = {digit * 7 for digit in digits}
B = {digit * 9 for digit in digits}
print("Set A:", A)
print("Set B:", B)
```

```python
union = A.union(B)
print("\nUnion of A and B:", union)
```

```python
intersection = A.intersection(B)
print("Intersection of A and B:", intersection)
```

```python
A_difference_B = A.difference(B)
B_difference_A = B.difference(A)
print("A - B:", A_difference_B)
print("B - A:", B_difference_A)
```

```python
symmetric = A.symmetric_difference(B)
print("Symmetric Difference:", symmetric)
```

```python
print("\nIs A a subset of B?", A.issubset(B))
print("Is B a superset of A?", B.issuperset(A))
```

```python
X = int(input("\nEnter a value to remove from Set A: "))
A.discard(X)
print("Set A after discard:", A)
```

```python
my_dict = {
"name": "Atishay",
"roll_no": "1024170358",
"branch": "CSE",
"age": 21,
"city": "Patiala"
}
print("Original Dictionary:")
print(my_dict)
```

```python
my_dict["location"] = my_dict.pop("city")
print("\nAfter renaming city to location:")
print(my_dict)
```

```python
my_dict["cgpa"] = 7.0
print("\nAfter adding CGPA:")
print(my_dict)
```

```python
my_dict["age"] = my_dict["age"] + 1
print("\nAfter increasing age by 1:")
print(my_dict)
```

```python
dict_pop = my_dict.copy()
removed_branch = dict_pop.pop("branch")
print("\nAfter deleting branch using pop():")
print(dict_pop)
print("Value returned by pop():", removed_branch)
```

```python
dict_del = my_dict.copy()
del dict_del["branch"]
print("\nAfter deleting branch using del:")
print(dict_del)
```

```python
print("\nKey → Value:")
for key, value in my_dict.items():
    print(key, "→", value)
```

```python
print("\nChecking for email:")

if "email" in my_dict:
    print("Email:", my_dict["email"])
else:
    print("Email is not present in the dictionary.")
```
